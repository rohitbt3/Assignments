### 2.Imputation

#### There are several ways to impute the data 
#### fill with mean, median, mode, forward fill, backward fill or with zero's

```python
df_copy = df.copy()
```

```python
null_col = []
for i in df.columns.tolist():
    if(df[i].isnull().sum()!=0):
        null_col.append(i)

print("Choose the option you want to fill : \n")
for i in null_col:
    print(i)
print()
in_ = input("Enter the option as it is or copy from the list")

op = input(f""" Please choose the desired option to perform : 1. Mean
                                               2. Median
                                               3. Mode
                                               4. Forward Fill
                                               5. Backward Fill
                                               6. Zero's \n""")
match op:
    case "Mean":
        df_copy[in_].fillna(np.round(df_copy[in_].mean(),5),inplace = True)

    case "Median":
        df_copy[in_].fillna(np.round(df_copy[in_].median(),5),inplace = True)

    case "Mode":
        df_copy[in_].fillna(np.round(df_copy[in_].mode()[0],5),inplace = True)
        
    case "Forward Fill":
        df_copy[in_].fillna(method = "ffill",inplace = True)

    case "Backward Fill":
        df_copy[in_].fillna(method = "bfill",inplace = True)

    case "Zero's":
        df_copy[in_].fillna(0,inplace = True)

    case _:
        print("Choose a valid option")

print("\n"*2)

df_copy.isnull().sum()
```

## Outlier Detection

#### Below is the code for individual column outlier detection

```python
print(f"""Choose one option :  

{", ".join(map(lambda x : str(x),df.columns.tolist()))} """)
print()
inp = input("Enter the option here : ")
print()
q1 = df[inp].quantile(0.25)
q2 = df[inp].median()
q3 = df[inp].quantile(0.75)
iqr = q3 - q1
up_whisker = q3 + 1.5 * iqr
low_whisker = q1 - 1.5 * iqr
outliers = []
for i in df[inp]:
    if i < low_whisker or i > up_whisker:
        outliers.append(i)

print(f"{"IQR / Outliers":>25}")
print()
print(f"{"Measures":<20}    {"Values"}")
print(f"{"Q1":<20} : {q1}")
print(f"{"Q3":<20} : {q3}")
print(f"{"IQR":<21}: {q3-q1}")
print(f"{"Upper Whisker":<20} : {up_whisker}")
print(f"{"Lower Whisker":<20} : {low_whisker}")
print()
print(f"{"Total Outliers are":<21}: {len(outliers)}\n")
print(f"Outliers are : {", ".join(map(lambda x : str(x),np.round(df[(df[inp] > up_whisker) | (df[inp] < low_whisker)][inp].tolist(),5)))}")
```

## 2.	Statistical Summary:

```python
def stats_in_one_go(inp):
    
    print()
    print(f"{"Stats for":>25} {inp.upper()}")
    print()
    print("Measures of Central Tendency".center(50),"\n")
    print(f"{"Measures":<20}   {"Values":<10}")
    print(f"{"Mean":<20} : {np.round(df[inp].mean(),6):<10}")
    print(f"{"Median":<20} : {np.round(df[inp].median()):<10}")
    
    if(len(df[inp].mode())>5):
        print(f"{"Mode":<20} :Too many values for Mode, {len(df[inp].mode())} unique values")

    else:
        print(f"{"Mode":<20} : {", ".join(map(lambda x : str(x), list(df[inp].mode()))):<10}")

    print()
    print("Measures of Dispersion".center(50))
    print(f"{"Measures":<20}    {"Values"}")
    print(f"{"Variance":<20} :  {np.round(df[inp].var(),3)}")
    print(f"{"Standard Deviation":<20} :  {np.round(df[inp].std(),3)}")
    print()
    print("Skewness and Kurtosis".center(50))
    print(f"{"Measures":<20}    {"Values"}")
    print(f"{"Skewness":<20} :  {np.round(df[inp].skew(),3)}")
    print()
    if(df[inp].skew() == 0):
        print("Observation : Normally Distribution")
    elif(df[inp].skew() > 0):
        print("Observation : Right Skewed")
    else:
        print("Observation : Left Skewed")

    print()
    print(f"{"Measures":<20}    {"Values"}")
    print(f"{"Kurtosis":<20} :  {np.round(df[inp].kurt(),3)}")
    print()
    if(df[inp].kurt() > 0):
        print("Observation : Leptokurtic")
    elif(df[inp].kurt() < 0):
        print("Observation : Platykurtic")
    else:
        print("Observation : Symmetrical")
```

```python
dtypes = list(df.select_dtypes(include = "number").columns)
print("Select an option")
print()
for i in dtypes:
    print(i)
print()
inp = input("Enter the name exactly or you can copy that, cause of your indolence : ")
stats_in_one_go(inp)
```
