## Estimation And Confidence Intervals

### Importing the Packages

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Given data

```python
# The durability of each print-head (in millions of characters) was recorded 
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
data = pd.DataFrame(data,columns = ["value"])
```

#### Mean and Standard Deviation

```python
mean_ = np.round(data["value"].mean(),3)
std_ = np.round(data["value"].std(),3)
```

## Confidence Interval

#### it is a range estimate, saying chances or occurrence of the values likely to be high in certain range
#### Mainly involves two distributions. They are normal,t - distributions.
#### Confidence Interval (CI) = [ x_bar + delta , x_bar - delta]
#### delta = z * (s / sqrt(n))
#### where x_bar is sample man
#### z is a z - score
#### s is a sample standard deviation

## a. Build 99% Confidence Interval Using Sample Standard Deviation
### Using existing methods

### When the population std is unknown and the sample size is <30. we go for T - Distribution


```python
# Given Data
# confidence interval = 0.99
# significance level = 0.01
n = len(data)
delta = std_/np.sqrt(len(data))
ci99 = stats.t.interval(confidence = 0.99, df = n - 1, loc = mean_,scale = delta)
np.round(ci99,3).tolist()
```

### b. Build 99% Confidence Interval Using Known Population Standard Deviation sigma = 0.2

### When the population std is known even the size is less than 30, normal distribution can be performed

```python
# Given Data
# confidence interval = 0.99
# significance level = 0.01
n = len(data)
pop_std = 0.2
ci99 = stats.norm.interval(confidence = 0.99,loc = mean_,scale = pop_std/np.sqrt(n))
np.round(ci99,3).tolist()
```

### When the population std is unknown, we go for t - distribution, 
### The main thing is need to find out t - critical value, it happens to be beyond the given percentage.
### like when the confidence value is 99% , the significance value is 1% which in turn divide in half, the tails of the curve determines the critical values on both ends

```python
# for 0.99 confidence interval 
# 0.01 is the significancs level
t_critical = 0
ci = float(input("Enter the value that you want to build a confidence interval, it has to between 0 - 100 : "))
ci = ci / 100
alpha = np.round((1 - ci)/2,3)
df_ = len(data)-1
t_critical = stats.t.ppf(1 - alpha,df = df_)
print(f"Confidence interval {ci*100} = Range will be in between {np.round([mean_ - t_critical * (std_/np.sqrt(len(data))) , mean_ + t_critical * (std_/np.sqrt(len(data)))],3)}")
```

### When the population standard deviation is known(which is rare), we gor for normal distribution

```python
# for 0.99 confidence interval 
# 0.01 is the significancs level
z_critical = 0
pop_std = float(input("Whats the population standard deviation : "))
ci = int(input("Enter the value that you want to build a confidence interval, it has to between 0 - 100 : "))
ci = ci / 100
alpha = np.round((1 - ci)/2,3)
z_critical = stats.norm.ppf(1 - alpha)
print(f"ci_{ci*100} = {np.round([mean_ - z_critical * (pop_std/np.sqrt(len(data))) , mean_ + z_critical * (pop_std/np.sqrt(len(data)))],3)}")
```
