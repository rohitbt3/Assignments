## Exercise - 1 : Prime Numbers

```python
x = int(input("Enter a number to check whether it is prime or not : "))
c = 0
for i in range(1,x+1):
    if(x % i == 0):      # a prime number is , a number divides by 1 and itself !!
        c = c + 1
print()
if(c == 2):                 
    print(f"{x} is a prime number")
else:
    print(f"{x} is not a prime number")
```

## Exercise 3: Squares of Even/Odd Numbers

```python
print(f"""Choose which one is it - 1.Odd
                         2.Even""")
in_ = int(input("Choose the option (number only) : "))
start = int(input("Enter the intial value for the range : "))
end = int(input("Enter the final value for the range : "))
print("The square values are :")
if(in_ == 1):
    for i in range(start,end):
        if(i % 2 != 0):                         #when a number divides by 2, it produces a reminder for odd number and none for even number
            print(i**2, end = " ")
else:
    for i in range(start,end):
        if(i % 2 == 0):
            print(i**2, end = " ")
```

## Exercise 4: Word counter

```python
import re
in_ = input("Enter the text : ")
delimiter = [",", "/", "-", "_", "!", ";", ":", "~", ".", "&"," ","?"]
for delimiter_ in delimiter:
   in_ =  in_.replace(delimiter_,",")                        # given text having any symbol will replace with ",". replace is the method
l = in_.split(",")                                           # after replacing , split will divide the word based on the ",", gives you a list of items well divided
for i in l:                                                   
    if(i == ""):                                             # excess symbol will be prited "" like this after getting converted. so removing it
        l.remove(i)
d={}
for i in l:
    if(i in d):
        d[i] = d[i] + 1                                      #assiging to dictionary
    else:
        d[i] = 1
print()
for key,values in d.items():
    print(f"{key} : {values}")
```
